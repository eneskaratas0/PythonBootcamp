```python
class Musician():

    def __init__(self,name):
        self.name = name
        print("musician class")

    def test(self):
        print("test1")

    def test1(sefl):
        print("test2")
```

```python
enes = Musician("Enes")
```

```python
enes.name
```

```python
enes.test()
```

```python
enes.test1()
```

```python
class MusicianPlus(Musician):

    def __init__(self,name):
        Musician.__init__(self,name)
        print("musician plus")

    def test3(self):
        print("test3")

    #override
    def test1(self):
        print("test1 test1 test1")
```

```python
class MusicianPlus(Musician):
    def __init__(self, name):
        # EN PROFESYONEL KULLANIM: 
        # super() otomatik olarak Musician'ı bulur ve 'name' değişkenini fırlatır
        # Üstelik super() kullandığında self parametresini elle yazmana gerek kalmaz, Python onu arka planda otomatik halleder:
        super().__init__(name)
        print("musician plus")

    def test3(self):
        print("test3")

    #override
    def test1(self):
        print("test1 test1 test1")
```

```python
atlas = MusicianPlus("Atlas")
```

```python
atlas.test()
```

```python
atlas.test1()
```

```python
atlas.name = "Atlas Samancioglu"
```

```python
atlas.name
```

```python
atlas.test3()
```

```python
atlas.test1()
```

```python
class Banana():

    def __init__(self,name):
        self.name = name

    def info(self):
        return f"100 calories {self.name}"

class Apple():

    def __init__(self,name):
        self.name = name

    def info(self):
        return f"150 calories {self.name}"
```

```python
banana = Banana("banana")
```

```python
banana.info()
```

```python
apple = Apple("apple")
```

```python
apple.info()
```

```python
fruitList = [banana,apple]
```

```python
for fruit in fruitList:
    print(fruit.info())
```

```python
class Phone():

    def __init__(self,name,price):
        self.name = name
        self.__price = price

    def info(self):
        print(f"{self.name} price is: {self.__price}")

    def changePrice(self,price):
        self.__price = price
```

```python
iphone = Phone("Iphone 14",500)
```

```python
iphone.info()
```

```python
iphone.price = 400
```

```python
iphone.info()
```

```python
iphone.__price = 600
```

```python
iphone.changePrice(300)
```

```python
iphone.info()
```

```python
from abc import ABC, abstractmethod
```

```python
class Car(ABC):

    @abstractmethod
    def maxSpeed(self):
        pass
```

```python
class Tesla(Car):

    def maxSpeed(self):
        print("200km")
```

```python
tesla = Tesla()
```

```python
tesla.maxSpeed()
```

```python
class Mercedes(Car):

    def maxSpeed(self):
        print("300")
```

```python
mercedes = Mercedes()
```

```python
mercedes.maxSpeed()
```

```python
class Fruit():

    def __init__(self,name,calories):
        self.name = name
        self.calories = calories

    def __str__(self):
        return f"{self.name}: {self.calories}"

    def __len__(self):
        return self.calories
```

```python
myFruit = Fruit("banana",150)
```

```python
myFruit.calories
```

```python
print(myFruit)
```

```python
len(myFruit)
```

```python
class Train():

    def __init__(self,name):
        self.name = name

    def __getitem__(self,key):
        if key == "a":
            return self.name
        else:
            return "Not Found"
```

```python
myTrain = Train("myTrain")
```

```python
myTrain["a"]
```

```python
myTrain["b"]
```
