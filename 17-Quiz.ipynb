```python
class AIModel():

    def __init__(self,model_adi,dogruluk_skorlari):
        self.model_adi = model_adi
        self.dogruluk_skorlari = dogruluk_skorlari
```

```python
class NeuralNetwork(AIModel):

    def __init__(self,model_adi,dogruluk_skorlari):
        super().__init__(model_adi,dogruluk_skorlari)
        print(f"{self.model_adi} modeli başarıyla hafızaya yüklendi.")

    def filtrele_ve_donustur(self):
        basarili_skorlar = filter(lambda x : x > 0.70, self.dogruluk_skorlari)

        yuzdelik_skorlar = map(lambda x: x*100, basarili_skorlar)

        return list(yuzdelik_skorlar)
```

```python
test_skorlari = [0.55, 0.75, 0.42, 0.88, 0.92, 0.68]
```

```python
modelimiz = NeuralNetwork("DeepMind-V2", test_skorlari)
```

```python
temiz_yuzdeler = modelimiz.filtrele_ve_donustur()
```

```python
print(f"0.70 üzerindeki başarılı skorların yüzdelik halleri: {temiz_yuzdeler}")
```

```python
class Ogrenci():
    
    def __init__(self,isim,sinavNotu):
        self.isim = isim
        self.__sinavNotu = sinavNotu
    
    def notuGoster(self):
        print(f"{self.isim} sınav notu: {self.__sinavNotu}")
```
