```python
#python 3.10
def process_value(value: int | str) -> str:
    if isinstance(value, int): # eger verilen deger int sa
        return f"processed integer: {value}"
    else:
        return f"processed string: {value}"
```

```python
process_value(50)
```

```python
process_value("ada")
```

```python
#python 3.9
from typing import List

def  sum_list(numbers: List[int]) -> int:
    return sum(numbers)
```

```python
numbers = [1,2,3,4]
sum_list(numbers)
```
