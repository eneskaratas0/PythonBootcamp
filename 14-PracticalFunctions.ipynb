```python
def controlString(string):
    return "Enes" in string
```

```python
myStringList = ["Enes","Atil Samancioglu", "Bahadir","Ali"]
```

```python
list(map(controlString,myStringList))
```

```python
#Filter
```

```python
list(filter(controlString,myStringList))
```

```python
#GLOBAL
myString = "ENES"

def myFunction():
    # ENCLOSING
    myString = "ATIL"

    def myFunction2():
        #LOCAL
        myString = "ESMA"
        print(myString)
    myFunction2()
```
