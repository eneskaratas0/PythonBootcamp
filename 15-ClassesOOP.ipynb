```python
class Dog():

    year = 7

    def __init__(self,age=5):
        self.age = age
        self.dogHumanAge = age * self.year

    def humanAge(self):
        return self.age * self.year # Dog.year -> self.year
```

```python
myDog = Dog(3)
```

```python
myDog.age
```
